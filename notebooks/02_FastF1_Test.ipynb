```python
import fastf1
from pathlib import Path

# Set up the cache folder
cache_path = Path("../cache")
fastf1.Cache.enable_cache(cache_path)

print("✅ FastF1 cache enabled successfully!")
print("Cache location:", cache_path.resolve())
```

```python
import fastf1

# Load the 2025 Australian Grand Prix race session
session = fastf1.get_session(2025, "Australian Grand Prix", "R")

print("Downloading session data...")
session.load()

print("✅ Session loaded successfully!")
```

```python
print("Event Name:", session.event['EventName'])
print("Year:", session.event['EventDate'].year)
print("Session:", session.name)
print("Drivers:", len(session.drivers))
```
